We apply the dynamical shake-up model to explain the super Earth-cold Jupiter relations and investigate the most possible configurations of those planetary systems hosting both cold Jupiter and super Earth. This model considers the stage of gas depletion after the formation of a Jupiter-like planet at a long distance from the host star when the inner embryos suffer the effect of secular resonance combined with Type I damping. During that stage, the eccentricities of the embryos are excited and they have a good chance to closely encounter other embryos, which leads to not only orbit crossing but also collisions, merger and scatterings. Several collisions could cause the formation of super Earths. At the same time, they migrate towards the star due to the damping caused by the residual disk until the disk is completely depleted, which could explain the relatively close location and the mearly circular orbit of the super Earth. 

The secular resonance is key to the formation of the super Earth. With a finite eccentricity, the cold Jupiter $j$ (with mas $m_j$, semimajor axis $a_j$, mean motion $n_j$, eccentricity $e_j$, and longitudes of periastron $\omega_j$) exert a secular perturbation on the inner embryos $i$ (with mass $m_i$, semimajor axis $a_i$, mean motion $n_i$, eccentricity $e_i$, and longitudes of periastron $\omega_i$), resulting in the precession of their longitudes of periastron and the modulation of their eccentricities. The precession is at a rate

\begin{equation}
\frac{d\omega_{i,j}}{dt} = \frac{1}{t_{i,j}}[1+C_{i,j}(\frac{e_j}{e_i}cos(\omega_i-\omega_j)]
\end{equation}
where 

\begin{equation}
t_{i,j}^{-1} = \frac{Gm_j\alpha b_{3/2}^{(1)}}{4n_i a_i^2max(a_i,a_j)} \equiv g_{i,j} 
\end{equation}

is a secular part of the precession of planet $i$ introduced by planet $j$, $\alpha$ = min($a_i$,$a_j$)/ max($a_i$,$a_j$) $C_{i,j} \equiv b_{3/2}^{(2)}/b_{3/2}^{(1)}$, where $b_s^(j)$ is the Laplace coefficient and it can be estimated as[Solar System Dynamics]: 

\begin{equation}
     b_s^{(k)}(\alpha) = \frac{s(s+1)...(s+k-1)}{1\cdot 2\cdot ... \cdot k} \alpha^k  \times [1 + \frac{s(s+k)}{1(k+1)} \alpha^2 \nonumber
     + \frac{s(s+1)(s+k)(s+k+1)}{1 \cdot 2(k+1)(k+2)}\alpha^4+...]
\end{equation}

As we focus on the orbital evolution of the embryos interior to the cold Jupiter, only accept the first ten terms of this expanding are taken into account. Noted that the effects on the cold Jupiter caused by the gravitational interaction with the embryos are so slight that we neglect it. 

Under the effect of the gravitational potential of the disk, the embryos and gas giants also undergo precessions. The initial apsidal precession rate of the embryos is 

\begin{equation}
    g_{i,disk} \equiv \frac{d\omega_{i,disk}}{dt} \sim -Z_k[\frac{\pi G\Sigma_{init}(a_i) }{n_i a_i}]
\end{equation}

where $\Sigma_{init} = \Sigma_0(\frac{a}{1AU})^{-k}$ is the initial surface density of the gas at distance $a$ and $\Sigma_0 = 50g cm^{-2}$, and $Z_k$ is a constant. Assuming a minmum-mass disk, $Z_{3/2} = 1.094$. (How could I obtain the value of Zk for k=1.01?)

During the depletion of the disk, the rate ($g'_{i,disk}$) changes under the power law,

\begin{equation}
    g'_{i,disk}=g_{i,disk}\rm{exp} (-\frac{t}{\tau_n})
\end{equation}
where $\tau_n$ is the depletion timescale of the gas disk

Once the precession rate of the embryos match that of the cold Jupiter, they will experience the secular resonance in which the phase ($\omega_i -\omega_j$) is locked. Therefore, the angular momentum is monotonically transferred from the embryos to the cold Jupiter and the eccentricity of the embryos will be excited. The resonance condition should be

\begin{equation}
    g_{i,j} + g'_{i,disk} = g'_{j,disk}  
\end{equation}

Noted that precession caused by the disk is much larger than that introduced by the giant planet everywhere in the minimum-mass nebula. Therefore, only when the disk is under depletion, can the maginitude of these precession rates be comparable to each other and then can the secular resonance occur. The time ($t_R$) when the secular resonance passes $a AU$ can be calculated as 

\begin{equation}
    u_R \equiv exp(-\frac{t_R}{\tau_n}) = \frac{g_{i,j}}{g_{j,disk}-g_{i,disk}}
\end{equation}



## Gap around the cold Jupiter

When the cold Jupiter gets massive enough, a gap is formed with a radial extend between $R_in$ and $R_out$. ((Goldreich \& Tremaine 1980))

\begin{equation}
    R_{\rm in} = a_j(1-e_j)(1 -(\frac{M_j}{3M_*})^{1/3}); \nonumber 
    R_{\rm out} = a_j(1+e_j)(1 +(\frac{M_j}{3M_*})^{1/3});
\end{equation}

Surrounded by a gap, the apsidal precession rate of the cold Jupiter caused by the disk is

\begin{equation}
    g_{j,disk}=\frac{d\omega_j}{dt} \sim (\frac{\pi G \Sigma_{init}(a_j)}{n_j a_j})\Sigma_{l=1}^{\infty}l(2l+1)A_l \nonumber 
     \times [\frac{(R_{\rm in}/a_j)^{2l+2-k}}{2l+2-k}+\frac{(a_j/R_{\rm out})^{2l-1+k}}{2l-1+k}]
\end{equation}
where $A_l = [(2l)!/2^{2l}/(l!)^2]^2$. During the disk depletion, the rate decrease as $g'_{j,disk} = g_{j,disk}exp(-t/\tau_n)$.

The change of the disk structure could also influence the precession rate of the interior embryos, it will be modified depending on the different locations to the gap. If the embryos are interior to the inner edge of the gap, the precession rate introduced by the disk is

\begin{equation}
    g_{i,disk} \equiv \frac{d\omega_{i,disk}}{dt} \sim -\frac{\pi G\Sigma_{init}(a_i) }{n_i a_i}\times(Z_k+ \nonumber 
    \Sigma_{l=1}^{\infty} l(2l+1)A_l  \frac{(a_i/R_{\rm in})^{2l-1+k}}{2l-1+k})
\end{equation}

where $Z_k$ is the same constant in equation (\ref{eq:disk_i}), and $A_l$ is described in equation (\ref{eq:disk_j_gap}). For embryos located in the gap range, its precession rate caused by the disk has the same functional form as equation (\ref{eq:disk_j_gap}), evaluated at $a_i$. As for embryos exterior to the outer edge of the gap, which we do not take into consideration in our numerical simulation, the rate is

\begin{equation}
    g_{i,disk} \equiv \frac{d\omega_{i,disk}}{dt} \sim -\frac{\pi G\Sigma_{init}(a_i) }{n_i a_i}\times(Z_k + \nonumber  \Sigma_{l=1}^{\infty}  l(2l+1)A_l  \frac{(R_{\rm out}/a_i)^{2l+2-k}}{2l+2-k})
\end{equation}

```python
def resonant_t_gap(a_p,a_p2,e_p,m_p,mstar,dens0,tdep,di,zk,s,j):
    import math
    a_1 = np.zeros(len(a_p2))
    a_2 = np.zeros(len(a_p2))
    for nn,aa in enumerate(a_p2):
        if aa <= a_p:
            a_1[nn] = aa
            a_2 [nn]= a_p
        else:
            a_1[nn] = a_p
            a_2[nn] = aa
    alpha = a_1/a_2

    n_c = np.sqrt(mstar / a_p2**3)
    n_b = np.sqrt(mstar / a_p**3)
    dens0_c = dens0*a_p2**(-di)
    dens0_b = dens0*a_p**(-di)
    def Lap_coef(s,j,alpha):
        term = pow(alpha,j)
        for nn in range(j):
            term = term * (s+nn)/(nn+1)

        sum_term = 1
        sum_subterm = 1
        for nn in range(1,10):
            sum_subterm *= (s+nn-1)*(s+j+nn-1)/(nn*(j+nn))
            sum_term += sum_subterm * alpha**(2*nn)
        return term*(sum_term)
    b_sj = Lap_coef(s,j,alpha)


    A=np.zeros(30)
    for jj in np.arange(0,30,1):
        l=jj+1
        A[jj] = (math.factorial(2*l)/(2)**(2*l)/( math.factorial(l))**2)**2

    ain = a_p*(1-e_p)*(1 - pow(m_p/(3*mstar),1/3));
    aout = a_p*(1+e_p)*(1 + pow(m_p/(3*mstar),1/3));

    sum_term =0
    for l in np.arange(1,11,1):
        sum_term+=l*(2*l+1)*A[l-1]*(pow(ain/a_p,2*l+2-di)/(2*l+2-di)+pow(a_p/aout,2*l-1+di)/(2*l-1+di))

    gb_disk = (math.pi*dens0_b/(n_b*a_p))*sum_term

    gi_disk = np.zeros(len(a_p2))
    for nn,aa in enumerate(a_p2):
        n_i = np.sqrt(mstar / aa**3)
        dens0_c= dens0*aa**(-di)
        if aa < ain:
            nt_sum=0
            for l in np.arange(1,31,1):
                nt = l*(2*l+1)*A[l-1]*(pow(aa/ain,2*l-1+di)/(2*l-1+di))
                nt_sum += nt;
            gi_disk[nn] = - (zk+nt_sum)*math.pi*(dens0_c/(n_i*aa))
        elif aa > aout:
            nt_sum=0
            for l in np.arange(1,31,1):
                nt = l*(2*l+1)*A[l-1]*(pow(aout/aa,2*l+2-di)/(2*l+2-di))
                nt_sum += nt;
            gi_disk[nn] = - (zk+nt_sum)*math.pi*(dens0_c/(n_i*aa))
        else:
            sum_term2 =0
            for l in np.arange(1,31,1):
                sum_term2 += l*(2*l+1)*A[l-1]*(pow(ain/aa,2*l+2-di)/(2*l+2-di)+pow(aa/aout,2*l-1+di)/(2*l-1+di))
            gi_disk[nn] = (math.pi*dens0_c/(n_i*aa))*sum_term2
    g_cb = m_p * alpha * Lap_coef(s,j,alpha)/(4*n_c*a_p2**2*a_2)
    u_R = (g_cb)/(gb_disk - gi_disk)
    t=np.zeros(len(u_R))
    for nn,uu in enumerate(u_R):
        if 0 < uu <= 1:
            t[nn] = - tdep*np.log(uu)
        else:
            t[nn] = np.nan
    return t
```

```python
def resonant_t_nogap(a_p,a_p2,e_p,m_p,mstar,dens0,tdep,di,zk,s,j):
    import math
    a_1 = np.zeros(len(a_p2))
    a_2 = np.zeros(len(a_p2))
    for nn,aa in enumerate(a_p2):
        if aa <= a_p:
            a_1[nn] = aa
            a_2 [nn]= a_p
        else:
            a_1[nn] = a_p
            a_2[nn] = aa
    alpha = a_1/a_2
    n_c = np.sqrt(mstar / a_p2**3)
    n_b = np.sqrt(mstar / a_p**3)
    dens0_c = dens0*a_p2**(-di)
    dens0_b = dens0*a_p**(-di)
    def Lap_coef(s,j,alpha):
        term = pow(alpha,j)
        for nn in range(j):
            term = term * (s+nn)/(nn+1)

        sum_term = 1
        sum_subterm = 1
        for nn in range(1,10):
            sum_subterm *= (s+nn-1)*(s+j+nn-1)/(nn*(j+nn))
            sum_term += sum_subterm * alpha**(2*nn)
        return term*(sum_term)
    gb_disk = -zk*(math.pi*dens0_b)/(n_b*a_p)
    gc_disk = -zk*(math.pi*dens0_c)/(n_c*a_p2)
    g_cb = m_p * alpha * Lap_coef(s,j,alpha)/(4*n_c*a_p2**2*a_2)
    u_R = (g_cb)/(gb_disk - gc_disk)
    t=np.zeros(len(u_R))
    for nn,uu in enumerate(u_R):
        if 0 < uu <= 1:
            t[nn] = - tdep*np.log(uu)
        else:
            t[nn] = np.nan
    return t
```

```python
def resonant_t_nodisk(a_p,a_p2,e_p,m_p,mstar,dens0,tdep,di,zk,s,j):
    return np.nan*np.ones_like(a_p2)
```

```python
def plot_orbit(fname,ax1,f_t,m_j,mstar,dens0,tdep,di,zk):
#     print(fname)
    sa = rebound.Simulationarchive(fname)

    sim = sa[0]

    t_array = np.zeros(len(sa))
    a_array = np.zeros([sim.N-1,len(sa)])
    e_array = np.zeros([sim.N-1,len(sa)])

    for i,sim1 in enumerate(sa):
        o = sim1.orbits(primary=sim1.particles[0])
        t_array[i] = sim1.t/1e6
        for n in range(sim1.N-1):
            a_array[n,i] = o[n].a
            e_array[n,i] = o[n].e

    for n in range(sim.N-1):
        ax1.plot(t_array,a_array[n,:],c='C%d'%n)
        ax1.plot(t_array,a_array[n,:]*(1+e_array[n,:]),c='C%d'%n,linestyle='--')
        ax1.plot(t_array,a_array[n,:]*(1-e_array[n,:]),c='C%d'%n,linestyle='--')


    a_p2 = np.arange(0.1,a_array[0,0]*(1-e_array[0,0]),0.01)
    ax1.plot(f_t(a_array[0,0],a_p2,e_array[0,0],m_j,mstar,dens0,tdep,di,zk,3/2,1)/1e6,a_p2,c="gray",linewidth=5,alpha=0.5)
    
def fname_idx(path,prefix,np,aj,ej,mp,ap0):
    return path+'%s_nj_1_np2_%d_mstar_1.00_mj_1.00_aj_%.2f_ej_%.2f_mp2_%.2f_ap2_%.2f_ep2_0.00.bin'%(prefix,np,aj,ej,mp,ap0)
```

```python
def sigma_cgs_to_code(sigma_cgs):
    return sigma_cgs * 1.125e-7


def feeding_zone(mstar):
    if np.isclose(mstar, 1.0):
        return 0.5, 4.0, 0.8, 1.5, 2.5
    if np.isclose(mstar, 0.6):
        return 0.12, 1.0, 0.20, 0.37, 0.61
    raise ValueError('feeding_zone is tabulated in ref2 only for selected stellar masses')


def sr_u_curve(a_j, e_j, m_j, mstar, sigma1_cgs, alpha, zk):
    a_min, a_max, hz_min, hz_max, water_line = feeding_zone(mstar)
    a_embryo = np.arange(a_min, a_max, 0.002)
    dens_init = sigma_cgs_to_code(sigma1_cgs) * mstar
    t_r = resonant_t_nogap(a_j, a_embryo, e_j, m_j, mstar, dens_init, tdep, alpha, zk, 3/2, 1)
    u_r = np.exp(-t_r / tdep)
    return a_embryo, u_r, (hz_min, hz_max, water_line)


m_j = 1.0 * 9.545e-4
alpha = 1.0
# Ref2 tabulates Z_{3/2}; it does not give Z_k for alpha=1, so keep the same coefficient here.
a_j_values = [5.0, 10.0, 20.0, 30.0]
e_j_values = [0.0, 0.25, 0.5]
linestyles = ['-', '--', ':']
colors = ['C0', 'C1', 'C2', 'C3']

fig, axes = plt.subplots(2, 2, figsize=(10, 7), sharey='row')
for row, mstar_i in enumerate([1.0, 0.6]):
    for col, sigma1 in enumerate([6.0, 60.0]):
        ax = axes[row, col]
        a_min, a_max, hz_min, hz_max, water_line = feeding_zone(mstar_i)
        ax.axvspan(a_min, a_max, color='0.88', zorder=0)
        ax.axvspan(hz_min, hz_max, color='green', alpha=0.18, zorder=0)
        ax.axvline(water_line, color='C0', linewidth=1.0, alpha=0.5)

        for color, a_j_i in zip(colors, a_j_values):
            for ls, e_j_i in zip(linestyles, e_j_values):
                a_embryo, u_r, _ = sr_u_curve(a_j_i, e_j_i, m_j, mstar_i, sigma1, alpha, zk)
                ax.plot(a_embryo, u_r, color=color, linestyle=ls, linewidth=1.4)

        ax.set_yscale('log')
        ax.set_ylim(1e-4, 1)
        ax.set_xlim(a_min, a_max)
        ax.set_title(rf'$M_*={mstar_i:g}M_\odot$, $\Sigma_1={sigma1:g}\,g\,cm^{{-2}}$')
        ax.set_xlabel('a [AU]')
        ax.grid(True, which='both', alpha=0.25)

axes[0, 0].set_ylabel(r'$u_R = \exp(-t_R/\tau_n)$')
axes[1, 0].set_ylabel(r'$u_R = \exp(-t_R/\tau_n)$')

color_handles = [plt.Line2D([0], [0], color=color, label=rf'$a_j={a_j_i:g}$ AU') for color, a_j_i in zip(colors, a_j_values)]
style_handles = [plt.Line2D([0], [0], color='k', linestyle=ls, label=rf'$e_j={e_j_i:g}$') for ls, e_j_i in zip(linestyles, e_j_values)]
region_handles = [
    plt.Rectangle((0, 0), 1, 1, color='0.88', label='feeding zone'),
    plt.Rectangle((0, 0), 1, 1, color='green', alpha=0.18, label='HZ'),
    plt.Line2D([0], [0], color='C0', alpha=0.5, label='water line'),
]
axes[0, 1].legend(handles=color_handles + style_handles, loc='lower right', fontsize=8)
axes[1, 1].legend(handles=region_handles, loc='lower right', fontsize=8)

fig.suptitle('Analytical SR location from ref2 no-gap setup')
fig.tight_layout()

from pathlib import Path
Path('figure').mkdir(exist_ok=True)
fig.savefig('figure/analytical.png', dpi=300)
```
